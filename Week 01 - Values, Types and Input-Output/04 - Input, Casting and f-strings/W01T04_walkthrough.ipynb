## 2. The one rule that catches everyone

> **`input()` always gives you text. Always. Even when the person types a number.**

Run the next cell and type `23.7`.

````python
value = input("Value : ")

print("You typed:", value)
print("Its type is:", type(value))
````

So this would fail with a `TypeError` — Python received the *characters* `2`, `3`, `.`, `7`, not a number.

Uncomment the line and see for yourself.

````python
 print(value + 1)
````

## 3. Casting — converting a type

| Function | Turns something into |
|---|---|
| `int()` | whole number |
| `float()` | decimal number |
| `str()` | text |

````python
text = "20"
print(text, type(text))

as_number = int(text)
print(as_number, type(as_number))
````

Usually done in the same line as the input. Read it **inside out**: ask → get text → convert → store.

````python
value = float(input("Value : "))

print(value + 1)
print(type(value))
````

> **Use `float()` unless you are certain the number is whole.** `int("23.7")` fails with a `ValueError`.

## 4. f-strings

Put `f` before the quotes, then any variable in `{ }`.

````python
record_id = "R-004"
value = 23.7

print(f"Record {record_id} has value {value}")
````

You can calculate inside the braces too.

````python
limit = 20
print(f"Over the limit by {value - limit}")
````

Forgetting the `f` is the most common slip. This prints the braces themselves:

````python
print("Record {record_id}")
````

## 5. Formatting — making it look like a report

| Written | Shows | Means |
|---|---|---|
| `{value}` | `23.7` | as is |
| `{value:.2f}` | `23.70` | 2 decimal places |
| `{value:>10}` | `      23.7` | right-aligned in 10 spaces |
| `{value:>10.2f}` | `     23.70` | both |
| `{diff:+.1f}` | `+3.7` | always show the sign |

````python
percent = (value / limit) * 100

print(f"{percent}")
print(f"{percent:.2f}")
print(f"{percent:>12.2f}")
````

Right-alignment is what makes a report line up instead of looking like a mess. Compare these two.

````python
difference = value - limit

print("--- without alignment ---")
print(f"Value : {value}")
print(f"Limit : {limit}")
print(f"Diff  : {difference}")

print()
print("--- with alignment ---")
print(f"Value : {value:>10.2f}")
print(f"Limit : {limit:>10.2f}")
print(f"Diff  : {difference:>+10.2f}")
````

## Topic 4 Challenge

Ask the user for a hostname, GB used and GB total. Then print exactly this shape, aligned:

```
==============================
  DISK REPORT  -  srv-01
==============================
  Used        :      87.00 GB
  Total       :     120.00 GB
```

This is a rehearsal for the mini-project. Get it working now and the lab is straightforward.

````python
# your code here

server_name = input("Enter the server name: ")
used_storage_of_server = float(input("Enter the used storage of the server in GB: "))
total_storage_of_server = float(input("Enter the total storage of the server in GB: "))
available_storage_of_server = total_storage_of_server - used_storage_of_server
percentage_of_used_storage = (used_storage_of_server / total_storage_of_server) * 100

print("=" *30)
print(f"  DISK REPORT  -  {server_name}")
print("=" *30)
print(f"  Used        :     {used_storage_of_server:>6.2f} GB")
print(f"  Total       :    {total_storage_of_server:>7.2f} GB")
print(f"  Free        : {available_storage_of_server:>10.2f} GB")
print(f"  Percent Used:  {percentage_of_used_storage:>9.1f}%")
print("=" *30)



````

---
## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| `input("n: ") + 1` | `TypeError` | `float(input("n: ")) + 1` |
| `int(input())` when they type `23.7` | `ValueError` | use `float()` |
| `print("Value {v}")` | prints the braces | you forgot the `f` |
| `f"{value:.2}"` | odd result | you meant `.2f` |
