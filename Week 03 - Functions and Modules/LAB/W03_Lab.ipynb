### D5. Arguments are matched by position

2. Print `within_limit(87, 100)`, then `within_limit(100, 87)`.
3. Add a comment explaining why the two results are different.

**Expected output:** `True`, then `False`

````python
# D5

def within_limit(value, limit):
    """Return True if value is within limit, else False."""
    return value <= limit

print (within_limit(87, 100))
print (within_limit(100, 87))

# they are different because the first one is True and the second one is False. The first one is True because 87 is less than 100, while the second one is False because 100 is greater than 87.

````

### D6. A default value

1. Define `status_of(percent, warning_at=90)` that returns:
   - `"OVER LIMIT"` if `percent` is 100 or more
   - `"WARNING"` if `percent` is `warning_at` or more
   - `"OK"` otherwise
2. Print `status_of(95)`, then `status_of(85)`, then `status_of(85, 80)`.

**Expected output:** `WARNING`, `OK`, `WARNING`

````python


# D6
def status_of(percent, warning_at=90):
    if percent >= 100:
        return "OVER LIMIT"
    if percent >= warning_at:
        return "WARNING"
    else:
        return "OK"


status1 = status_of(95)
status2 = status_of(85)
status3 = status_of(85,80)
print(status1, status2, status3)
````

### D8. `*args` and `**kwargs`

a) Define `total(*numbers)` that returns the sum of all the numbers passed in. Print `total(1, 2, 3)`,
   `total(10, 20, 30, 40)` and `total()`.

b) Define `configure(**settings)` that loops over `settings.items()` and prints each one as `key = value`.
   Call it with `configure(host="10.0.0.1", port=8080)`.

**Expected output:**
```
6
100
0
host = 10.0.0.1
port = 8080
```

````python
# D8

def total(*numbers):
    """Return the sum of all numbers."""
    return sum(numbers)

def configure(**settings):
    """Print the settings."""
    for key, value in settings.items():
        print(f"{key} = {value}")

print(total(1, 2, 3))
print(total(10,20,30,40))
print(total())
configure(host="10.0.0.1", port=8080)
````

### D9. Local means local

````python
# D9

# i predict that the function will print "inside: OK" and then "outside: OK"
def make_status():
    result = "OK"
    print("inside:", result)

make_status()
print("outside:", result)

# because the variable result is defined inside the function make_status, it is not accessible outside of that function. Therefore, when we try to print result outside of the function, we get a NameError because result is not defined in that scope.
````

````python
# D10
def double(n):
    n = n * 2
    return n

number = 21
print(double(number))
print(number)

# becuase its not changing the value of number outside the function, it is only changing the value of n inside the function. Therefore, when we print number after calling double, it still has its original value of 21.
````

### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |
| `count_over()` | `0` |

````python
# C2

def count_over(*values, limit=100):
    """Count how many values are over the limit."""
    count = 0
    for value in values:
        if value > limit:
            count += 1
    return count

print(count_over(87, 120, 45, 101))
print(count_over(87, 120, 45, 101, limit=50))
print(count_over())
````
